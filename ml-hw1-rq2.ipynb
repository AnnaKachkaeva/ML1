```python
# =========================
# COMMON SETUP
# =========================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# Load dataset
DATA_PATH = "/kaggle/input/datasets/sharmajicoder/gaming-and-mental-health/gaming_mental_health_10M_40features.csv"
df = pd.read_csv(DATA_PATH)

# -------------------------
# Basic Cleaning
# -------------------------
df = df.dropna()

# Example target (modify if needed)
TARGET = "daily_gaming_hours"  # adjust to actual column name

# Encode target if categorical
le = LabelEncoder()
df[TARGET] = le.fit_transform(df[TARGET])

# Separate features
X = df.drop(columns=[TARGET])
y = df[TARGET]

# Encode categorical features
X = pd.get_dummies(X)

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

def evaluate_model(model):
    y_pred = model.predict(X_test)
    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average="weighted"),
        "Recall": recall_score(y_test, y_pred, average="weighted"),
        "F1": f1_score(y_test, y_pred, average="weighted")
    }
```

```python
# =========================
# RQ2: Advanced Models
# =========================
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier

models = {
    "Random Forest": RandomForestClassifier(criterion = 'entropy', random_state = 42),
    # "SVM": SVC(probability=True),
    # "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric="mlogloss")
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    metrics = evaluate_model(model)
    metrics["Model"] = name
    results.append(metrics)

df_results = pd.DataFrame(results)
df_results.to_csv("RQ2_table.csv", index=False)

plt.figure()
plt.barh(df_results["Model"], df_results["Accuracy"])
plt.title("Model Comparison")
plt.savefig("RQ2_figure.pdf")
```
